```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent

METADATA_PATH = PROJECT_ROOT / "metadata"
SPLITS_PATH = PROJECT_ROOT / "splits"

SPLITS_PATH.mkdir(exist_ok=True)

print("Metadata:", METADATA_PATH)
print("Splits:", SPLITS_PATH)
```

```python
# Load metadata

mvtec_metadata = pd.read_csv(
    METADATA_PATH / "mvtec_metadata.csv"
)

visa_metadata = pd.read_csv(
    METADATA_PATH / "visa_metadata.csv"
)

print("MVTec rows:", len(mvtec_metadata))
print("VisA rows:", len(visa_metadata))
```

```python
# MVTec unsupervised split

mvtec_unsupervised = mvtec_metadata.copy()

# Original MVTec training data is normal-only.
# Keep the official train/test structure.

mvtec_unsupervised["experiment"] = "unsupervised"

mvtec_unsupervised.to_csv(
    SPLITS_PATH / "mvtec_unsupervised.csv",
    index=False
)

print("MVTec unsupervised split saved.")
print(
    mvtec_unsupervised
    .groupby(["split", "label"])
    .size()
    .unstack(fill_value=0)
)
```

```python
# VisA unsupervised split

visa_unsupervised = visa_metadata.copy()

visa_unsupervised["experiment"] = "unsupervised"

visa_unsupervised.to_csv(
    SPLITS_PATH / "visa_unsupervised.csv",
    index=False
)

print("VisA unsupervised split saved.")
print(
    visa_unsupervised
    .groupby(["split", "label"])
    .size()
    .unstack(fill_value=0)
)
```
